### Based on the metadata, what is the difference between the 6 columns starting with 'C' and the 6 related columns that don't?

The six columns beginning with `C` contain career totals: at bats, hits, home runs, runs, runs batted in, and walks. The corresponding columns without `C` contain totals for the 1986 season only. See the [Hitters metadata](https://intro-stat-learning.github.io/ISLP/datasets/Hitters.html).

### On the training set, create pairwise scatterplots for each of these 6 columns with the 'Salary' variable.

```python
# First create a subset of the columns that we want to plot

subset = Train[['CAtBat', 'CHits', 'CHmRun', 'CRuns', 'CRBI', 'CWalks']]

# Initialize the plots before drawing them

fig, axes = subplots(nrows=2,
                     ncols=3,
                     figsize=(15, 10))
# Copy the helper function

def range_to_grid(i,Ncol):
    x=[]
    y=[]
    for n in range(Ncol**2):
        x.append(int(np.floor(n/Ncol)))
        y.append(n % Ncol)
        #print(n,x[n],y[n])
    return x[i],y[i]

# Plot the variables

for j in range(len(subset.columns)):
    axes[range_to_grid(j,3)[0],range_to_grid(j,3)[1]].plot(subset.iloc[:,j], Train['Salary'], 'o')
    axes[range_to_grid(j,3)[0],range_to_grid(j,3)[1]].set_xlabel(subset.columns[j])

    axes[range_to_grid(j,3)[0],range_to_grid(j,3)[1]].set_ylabel('Salary (thousands of dollars)')

fig.suptitle('Career Statistics versus Salary (Training Set)')
fig.tight_layout()
plt.show()
```

### Describe a situation where a variable could have missing values but this would not be reflected in the results of the 'describe' method.

Missing observations might be stored as a numeric placeholder such as -999 instead of `NaN`. In that case, `describe()` treats the placeholder as an observed value: it is included in the count and summary statistics, so the count does not reveal the missing data. Also, the default `describe()` for this mixed dataset excludes categorical columns, so missing values in those columns would not appear in its numeric summary.

### For our last exercise, we'll explore Hits and Walks relative to AtBat totals. 
- Use the sum function to calculuate the totals of each of these three variables for the 1986 season (on the training set). 
- Create a pie chart which shows total hits, total walks, and remaining total (neither) as percents of the At Bats total (on the training set). 

```python
TotHits = Train['Hits'].sum()
TotWalks = Train['Walks'].sum()
TotAtBat = Train['AtBat'].sum()

Labels = ['Hits', 'Walks', 'Neither']
Totals = [TotHits, TotWalks, TotAtBat-TotHits-TotWalks]
print(f'Total hits: {TotHits}, total walks: {TotWalks}, total at bats: {TotAtBat}')
```

```python
# Pie chart using the assignment's requested AtBat denominator.
fig, ax = plt.subplots(figsize=(7, 7))
ax.pie(Totals, labels=Labels, autopct='%1.1f%%', startangle=90)
ax.set_title('1986 Totals as Percentages of AtBat (Training Set)')
ax.axis('equal')
plt.show()
# Baseball note: walks are excluded from official at-bat totals, so this
# requested chart is an arithmetic exercise, not a partition of actual at bats.
```

### The previous two cells gave us totals across all players. For each player in the training set, calculate the Hits as a percent of AtBat and store it in a new variable called 'AVG'

```python
# Store AVG as a proportion so that the next cell's 0.25 and 0.31 cutoffs apply.
Train['AVG'] = Train['Hits'] / Train['AtBat']
Train[['Hits', 'AtBat', 'AVG']].head()
```

### Using 0.25 and 0.31 as the split points, create a new variable with three bins: high, medium, and low. 

```python
Train['AVG_bin'] = 'medium'
Train.loc[Train['AVG'] < 0.25, 'AVG_bin'] = 'low'
Train.loc[Train['AVG'] > 0.31, 'AVG_bin'] = 'high'
# Values exactly at either cutoff stay in the medium category.
```

### Create a bar chart that displays the number of players in each of the low, medium, and high categories (for the training data).

```python
counts = Train['AVG_bin'].value_counts()
print(counts)
counts.plot.bar(rot=0, figsize=(7, 5))
plt.title('1986 AVG (Training Set)')
plt.xlabel('AVG Category')
plt.ylabel('Number of Players')
plt.tight_layout()
plt.show()
```

Notice that the order of the bars will be medium, low, high. That's counterintuitive. We can reorder these quickly. 

```python
indexMap = ['low', 'medium', 'high']
reordered_list = Train['AVG_bin'].value_counts().reindex(indexMap, fill_value=0).tolist()
```

```python
plt.figure(figsize=(7, 5))
plt.bar(range(len(indexMap)), reordered_list)

plt.title("1986 AVG (Training Set)")
plt.xlabel('AVG Category')
plt.ylabel("Number of Players")

plt.xticks(range(len(indexMap)), indexMap)
plt.tight_layout()
plt.show()
```

### Did we use the depth method or width method for creating these bins? Explain.

This is width-based (value-based) binning rather than depth-based binning: we set cutoffs on the AVG scale at 0.25 and 0.31 instead of choosing cutoffs to put the same number of players in each bin. The resulting counts are 65 low, 114 medium, and 14 high, so the bins do not have equal depth. Strictly speaking, these are manually specified intervals, not equal-width intervals, because the supplied cutoffs do not divide the observed AVG range into three equal lengths.
